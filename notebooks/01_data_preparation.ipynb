# Notebook 01 — Data Preparation

Download and clean the OECD/NEA Phase-1 CHF benchmark dataset.

```python
import requestsimport pandas as pdimport numpy as npfrom pathlib import PathDATA_DIR = Path('../data')DATA_DIR.mkdir(parents=True, exist_ok=True)url = ('https://raw.githubusercontent.com/zhouzhouwen/'       'cutting-edge-artificial-intelligence-methods-for-prediction-of-critical-heat-flux/'       'main/phase1-chf-exercises-main%20(5input)/data/chf_public.csv')r = requests.get(url, timeout=60)with open(DATA_DIR / 'chf_public.csv', 'wb') as f:    f.write(r.content)print('Downloaded.')
```

```python
df = pd.read_csv(DATA_DIR / 'chf_public.csv', skiprows=[1])numeric = ['Tube Diameter', 'Heated Length', 'Pressure', 'Mass Flux',           'Outlet Quality', 'Inlet Subcooling', 'Inlet Temperature', 'CHF']for c in numeric:    df[c] = pd.to_numeric(df[c], errors='coerce')df_clean = df.dropna(subset=numeric).reset_index(drop=True)df_clean.to_parquet(DATA_DIR / 'chf_clean.parquet', index=False)print(f'Cleaned: {len(df_clean):,} rows')print(f'Reference IDs: {df_clean["Reference ID"].nunique()}')
```
